```python

# 1. Print

print("Hello, World!")

print("=" *60)
# 2. Variables

hostname = "Core-switch"
ip_address = "192.168.1.1"

print("=" * 60)
print("Welcome to Cisco Network Automation Class with Python")
print("=" * 60)
print(type(hostname))
print(f"WAN Device is a {hostname}")
print(f"IP Address is {ip_address}")

print("="*60)

# 3. Integer and Float

ssh_port = 22
ios_version = 15.2

print(type(ssh_port))
print(type(ios_version))

print("="*60)

# 4. None

a = None

print(a)
print(type(a))

print("="*60)

# 5. Type Conversion

ios_version = 15.2
ios_ver = str(ios_version)

print(type(ios_ver))

print("="*60)

# 6. String Strip and Split

raw_cli_data = "Ethernet0          172.16.151.56   YES NVRAM  up                    up \n"

clean_data = raw_cli_data.strip()
print(clean_data)

columns = clean_data.split()
print(columns)

print("="*60)

# 7. List and List Methods

print(type(columns))
print(len(columns))

columns.append("New Column")
print(columns)

columns.insert(2, "Inserted Column")
print(columns)

columns.remove("Inserted Column")
print(columns)

print("="*60)

# 8. List Indexing

intf_name = columns[0]
ip_address = columns[1]
status = columns[4]
protocol = columns[5]

print(f"Interface Name: {intf_name}")
print(f"IP Address: {ip_address}")
print(f"Status: {status}")
print(f"Protocol: {protocol}")

print("="*60)


# 9. Nested List

devices = [
    "Core-RTR-01",
    "Core-RTR-02",
    "Core-SW-01",
    "Core-SW-02",
    ["Device1", "Device2"]
]

device_name = devices[4][0]

print(f"Device Name: {device_name}")

print("="*60)


# 10. Combining Lists

list1 = [1, 2, 3, 4, 5]
list2 = [6, 7, 8, 9, 10]

list3 = list1 + list2

print(list3)

print("="*60)

# 11. Append List

list2.append(list1)

print(list2)

print("="*60)

# 12. Remove

list1.remove(4)

print(list1)

print("="*60)

# 13. Pop

list1.pop(1)

print(list1)

print("="*60)

# 14. Tuple

tuple1 = (87.88, 64)

print(tuple1[1])

print("="*60)

# 15. Range

x = range(10)

print(x)

x1 = range(0, 10, 2)

print(list(x1))

print("="*60)

# 16. Dictionary

d = {
    "hostname": "core name",
    "ip_address": "140.12.52.114"
}

print(d)

print(d["hostname"])

print("="*60)


# 17. Nested Dictionary

d1 = {
    "hostname": "krishna",
    "ip_address": "144.12.522.55",
    "version": "9.5.2",

    "model": {
        "filter": "cisco",
        "version": "9.5.2"
    }
}

print(d1)

print(d1["ip_address"])

print(d1["model"])

print(d1["model"]["filter"])

print("="*60)


# 18. Network Device Dictionary

router = {
    "hostname": "Core-RTR-01",
    "ip_address": "192.168.1.1",
    "ssh_port": 22,
    "username": "admin",

    "model": {
        "vendor": "Cisco",
        "platform": "IOS XE",
        "version": "17.9"
    }
}

print(router)

print(router["hostname"])
print(router["ip_address"])
print(router["ssh_port"])
print(router["model"])
print(router["model"]["vendor"])
print(router["model"]["platform"])
print(router["model"]["version"])


print("="*60)

# 19. Network CLI Data Processing

raw_cli_data = """
Ethernet0          172.16.151.56   YES NVRAM  up                    up
"""

clean_data = raw_cli_data.strip()

columns = clean_data.split()

interface_name = columns[0]
ip_address = columns[1]
status = columns[4]
protocol = columns[5]

print(f"Interface Name: {interface_name}")
print(f"IP Address: {ip_address}")
print(f"Status: {status}")
print(f"Protocol: {protocol}")


print("="*60)
```
