```python
def check_interface_status(intf_name,admin_status,line_protocol):
    if admin_status == "up" and line_protocol == "up":
        return f"Interface {intf_name} is up and operational"
    elif admin_status == "up" and line_protocol == "down":
        return f"Interface {intf_name} is administratively up but line protocol is down"
    elif admin_status == "down" and line_protocol == "up":
        return f"Interface {intf_name} is administratively down but line protocol is up"   
    else:
        return f"Interface {intf_name} is down or not operational"

result = check_interface_status("Ethernet0","down","down")

print(result)
```
